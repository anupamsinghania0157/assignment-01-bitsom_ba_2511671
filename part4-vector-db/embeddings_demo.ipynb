```python
sentences = [
    # Cricket
    "Virat Kohli scored a century in the match",
    "The bowler took a wicket in the final over",
    "India won the cricket match by 5 runs",

    # Cooking
    "Add salt and spices while cooking vegetables",
    "Boil the pasta until it becomes soft",
    "Cooking requires patience and proper ingredients",

    # Cybersecurity
    "Use strong passwords to protect your accounts",
    "Hackers try to steal sensitive data from systems",
    "Install antivirus software to secure your computer",
    "Cybersecurity is important in the digital world"
]
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')

embeddings = model.encode(sentences)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

plt.figure(figsize=(8,6))
sns.heatmap(similarity_matrix, annot=False)
plt.title("Cosine Similarity Matrix")
plt.show()
```

```python
query = "The bowler took three wickets in one over"

query_embedding = model.encode([query])

similarities = cosine_similarity(query_embedding, embeddings)[0]

# Get top 2 similar sentences
top_indices = similarities.argsort()[-2:][::-1]

for idx in top_indices:
    print(sentences[idx], "-> Score:", similarities[idx])
```
